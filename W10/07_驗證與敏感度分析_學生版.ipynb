## 共用資料準備
以下在記憶體重建與既有教材相同的資料。執行後可以直接開始本單元；不會覆寫 CSV。


````python
# 匯入套件：numpy 建立隨機資料，pandas 管理表格，pyplot 繪圖
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# 固定種子，並維持呼叫順序，以重現原教材的 60 筆模擬資料
rng = np.random.default_rng(42)
n = 60
departments = ["資訊管理", "企業管理", "金融管理"]
# 每個字典鍵是一個欄位；對應的值是 60 名學生的資料
data = {
    "student_id": range(1, n + 1),
    "department": rng.choice(departments, n),
    "sleep_hours": rng.normal(6.7, 1.1, n).round(1),
    "screen_time_hours": rng.normal(5.0, 1.5, n).clip(1, 10).round(1),
    "study_hours": rng.normal(2.5, 1.0, n).clip(0.5, 6).round(1),
    "attendance_rate": rng.normal(88, 8, n).clip(60, 100).round(0),
}
df = pd.DataFrame(data)
# 成績依設定公式與隨機誤差建立；關係來自模擬規則，不能視為因果發現
noise = rng.normal(0, 5, n)
score = (50 + 3 * df["study_hours"] + 2 * df["sleep_hours"]
         - 1.2 * df["screen_time_hours"] + 0.15 * df["attendance_rate"] + noise)
df["quiz_score"] = score.clip(0, 100).round(0)
# 故意在計分後加入資料品質問題；無法從表格還原計分時原始值
df.loc[[5, 17], "sleep_hours"] = np.nan
df.loc[8, "screen_time_hours"] = 15.0
print("模擬資料形狀（列, 欄）：", df.shape)
````

````python
import numpy as np
import pandas as pd
from IPython.display import display

# 4 筆小資料；noise 故意設成已知，方便「先算對、再覆寫、再比對」
study  = np.array([2.0, 3.0, 1.0, 4.0])
sleep  = np.array([7.0, 6.0, 8.0, 5.0])
screen = np.array([4.0, 6.0, 2.0, 8.0])     # 計分當時的值 V
attend = np.array([90.0, 80.0, 95.0, 70.0])
noise  = np.array([1.0, -2.0, 0.5, -1.0])

df2 = pd.DataFrame({"student_id": [1, 2, 3, 4], "study_hours": study,
                    "sleep_hours": sleep, "screen_time_hours": screen,
                    "attendance_rate": attend})
df2["quiz_score"] = (50 + 3*df2["study_hours"] + 2*df2["sleep_hours"]
                     - 1.2*df2["screen_time_hours"]
                     + 0.15*df2["attendance_rate"] + noise).round(0)

# ★ 關鍵：計分「之後」才覆寫，跟你的教材同一個手法
df2.loc[1, "screen_time_hours"] = 15.0

# 用「目前欄位」把公式重算一遍（不含 noise）
base_now = (50 + 3*df2["study_hours"] + 2*df2["sleep_hours"]
            - 1.2*df2["screen_time_hours"] + 0.15*df2["attendance_rate"])
check = pd.DataFrame({"目前螢幕時間": df2["screen_time_hours"],
                      "表上分數": df2["quiz_score"],
                      "用目前欄位重算": base_now,
                      "表上分數扣除noise": df2["quiz_score"] - noise})
check["差異"] = check["用目前欄位重算"] - check["表上分數扣除noise"]
display(check)
````
