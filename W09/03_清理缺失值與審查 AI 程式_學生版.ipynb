````python
# 匯入套件：numpy 建立隨機資料，pandas 管理表格，pyplot 繪圖
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# 固定種子，並維持呼叫順序，以重現原教材的 60 筆模擬資料
rng = np.random.default_rng(42)
n = 60
departments = ["資訊管理", "企業管理", "金融管理"]
# 每個字典鍵是一個欄位；對應的值是 60 名學生的資料
data = {
    "student_id": range(1, n + 1),
    "department": rng.choice(departments, n),
    "sleep_hours": rng.normal(6.7, 1.1, n).round(1),
    "screen_time_hours": rng.normal(5.0, 1.5, n).clip(1, 10).round(1),
    "study_hours": rng.normal(2.5, 1.0, n).clip(0.5, 6).round(1),
    "attendance_rate": rng.normal(88, 8, n).clip(60, 100).round(0),
}
df = pd.DataFrame(data)
# 成績依設定公式與隨機誤差建立；關係來自模擬規則，不能視為因果發現
noise = rng.normal(0, 5, n)
score = (50 + 3 * df["study_hours"] + 2 * df["sleep_hours"]
         - 1.2 * df["screen_time_hours"] + 0.15 * df["attendance_rate"] + noise)
df["quiz_score"] = score.clip(0, 100).round(0)
# 故意在計分後加入資料品質問題；無法從表格還原計分時原始值
df.loc[[5, 17], "sleep_hours"] = np.nan
df.loc[8, "screen_time_hours"] = 15.0
print("模擬資料形狀（列, 欄）：", df.shape)
````

## 3.3 結果不符預期的除錯
**課堂提問：**原本想找小於 75 分，為何 `<= 75` 結果可能不同？先檢查預期、輸入、條件及中間結果。
**AI 提示詞：**請用三筆 74、75、76 分資料說明 `<` 與 `<=`；保留原資料、逐步列出結果並提供斷言。

用 74、75、76 說明 < 與 <=
核心：兩者只在「剛好等於門檻」的那一筆分開
門檻設 75（這題最關鍵的就是剛好 75 的 B）：

| student_id | quiz_score | `score < 75` | `score <= 75` |
|---|---|---|---|
| A | 74 | True | True |
| B | **75** | **False** | **True** ← 唯一差異 |
| C | 76 | False | False |
| **筆數**（`.sum()`） | | **1** | **2** |

一句話： < 是「嚴格低於」，<= 是「低於或等於」；當欄位裡存在剛好等於門檻的值時，<= 會比 < 多選中「等於門檻的那些筆」，一個都不多。


````python
# 建立邊界值 74、75、76，檢查條件而不是只看程式是否報錯
boundary = pd.DataFrame({"quiz_score": [74, 75, 76]})
less_than = boundary[boundary["quiz_score"] < 75]
less_or_equal = boundary[boundary["quiz_score"] <= 75]
print("小於 75：", less_than["quiz_score"].tolist())
print("小於等於 75：", less_or_equal["quiz_score"].tolist())
assert less_than["quiz_score"].tolist() == [74]
````
